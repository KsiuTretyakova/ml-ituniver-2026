````python
import pandas as pd

data = {
    "days": [1, 2, 3, 4, 5, 6, 7],
    "coins": [20, 21, 18, 12, 10, 1, 5]  # скільки в копілці
}

df = pd.DataFrame(data)
df
````

````python
import matplotlib.pyplot as plt

plt.scatter(df['days'], df['coins'], color='blue')
plt.title("Графік залежності")
plt.xlabel("days")
plt.ylabel("coins")
plt.show()
````

`y = WX + b`


````python
from sklearn.linear_model import LinearRegression

X = df[['days']]  # features - ознаки
y = df['coins']  # target - ціль

model = LinearRegression()
model.fit(X, y)

print("Вага (W):", model.coef_)
print("Зміщення (b):", model.intercept_)
````

````python
y_pred = model.predict(X)

plt.scatter(X, y, color='blue')  # крапочки
plt.plot(X, y_pred, color="red")  # лінія
plt.title("Лінійна регресія")
plt.xlabel("days")
plt.ylabel("coins")
plt.show()
````

````python
new_value = [[10]]
predicted_coins = model.predict(new_value)
print(f"Прогноз для {new_value[0][0]} днів: {predicted_coins[0]} монеток")
````
